# 🟥1.1 Derive an analytical solution to the regression problem. Use a vector form of the equation.

Аналитический метод — это способ решения задачи, при котором ответ находится по явной формуле за конечное число шагов (без итераций, без подбора параметров). Пример: корни квадратного уравнения по формуле 


Метод наименьших квадратов (МНК) — это критерий качества модели: выбираются такие коэффициенты, чтобы сумма квадратов ошибок была минимальной. МНК не является методом решения — это то, что мы хотим получить. Аналитическое решение МНК — это формула $\beta = (X^T X)^{-1} X^T y$

решить регрессионную задачу» означает «найти такие коэффициенты β, при которых ошибка минимальна». А ошибка — это функция потерь.

## Analytical solution to the linear regression problem

We have:
- $X$ — matrix of features (size $n \times m$)
- $y$ — vector of target values (size $n \times 1$)
- $\beta$ — vector of coefficients (size $m \times 1$)

The prediction is:
$$
\hat{y} = X\beta
$$

The loss function (sum of squared errors) is:
$$
L(\beta) = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 = (y - X\beta)^T (y - X\beta)
$$

### Step 1. Expand the loss function

$$
L(\beta) = (y^T - \beta^T X^T)(y - X\beta)
$$

$$
L(\beta) = y^T y - y^T X \beta - \beta^T X^T y + \beta^T X^T X \beta
$$

Note that $y^T X \beta$ is a scalar, so $y^T X \beta = (y^T X \beta)^T = \beta^T X^T y$. Therefore:

$$
L(\beta) = y^T y - 2\beta^T X^T y + \beta^T X^T X \beta
$$

### Step 2. Compute the gradient

The gradient of $L(\beta)$ with respect to $\beta$ is:

$$
\nabla L(\beta) = -2 X^T y + 2 X^T X \beta
$$

### Step 3. Set the gradient to zero (finding the minimum)

$$
-2 X^T y + 2 X^T X \beta = 0
$$

Divide by 2:

$$
- X^T y + X^T X \beta = 0
$$

$$
X^T X \beta = X^T y
$$

### Step 4. Solve for $\beta$

If $X^T X$ is invertible, multiply both sides by $(X^T X)^{-1}$:

$$
\boxed{\beta = (X^T X)^{-1} X^T y}
$$

This is the **analytical solution** (Ordinary Least Squares, OLS).

# 🟥1.2 What changes in the solution when L1 and L2 regularizations are added to the loss function.

**Problems without regularization:**
- If features are correlated, $X^T X$ can be non-invertible (no unique solution)
- Coefficients can become very large → overfitting
- The model is unstable (small changes in data cause large changes in predictions)

### 2. L2 Regularization (Ridge Regression)

**Adds a penalty on the sum of squared coefficients:**

$$
L_{\text{Ridge}}(\beta) = \underbrace{(y - X\beta)^T (y - X\beta)}_{\text{error}} + \lambda \underbrace{\sum_{j=1}^{m} \beta_j^2}_{\text{L2 penalty}}
$$

**What changes in the solution:**

The analytical solution becomes:

$$
\boxed{\beta = (X^T X + \lambda I)^{-1} X^T y}
$$

**Key properties of L2:**
- $X^T X + \lambda I$ is **always invertible** (even if $X^T X$ is singular)
- All coefficients are **reduced** (shrunk toward zero), but **none become exactly zero**
- Features with correlated effects get similar small weights
- $\lambda$ (lambda) is a hyperparameter that controls the strength of regularization


### 3. L1 Regularization (Lasso Regression)

**Adds a penalty on the sum of absolute coefficients:**

$$
L_{\text{Lasso}}(\beta) = \underbrace{(y - X\beta)^T (y - X\beta)}_{\text{error}} + \lambda \underbrace{\sum_{j=1}^{m} |\beta_j|}_{\text{L1 penalty}}
$$

**What changes in the solution:**

- **No closed-form analytical solution** (because $|\beta_j|$ is not differentiable at zero)
- The solution is found numerically (coordinate descent, gradient-based methods)
- Some coefficients become **exactly zero**

**Key properties of L1:**
- Performs **automatic feature selection** (zero coefficients = removed features)
- Produces sparse solutions (few non-zero coefficients)
- Useful when many features are irrelevant


# 🟥1.3 Explain why L1 regularization is often used to select features. Why are there many weights equal to 0 after the model is fit?

Отбор признаков (feature selection) — это процесс, при котором модель сама решает, какие входные данные важны, а какие нет.


L1 penalizes weights proportionally to their **absolute value** ($\lambda \sum |\beta_j|$).  
If a feature gives only a small benefit to the model, it's better to set its weight to **exactly zero** than to pay the penalty.

Geometric intuition

L1 constrains weights to a **diamond** (square rotated 45°). The diamond has **corners** on the axes. The optimal solution often lands at a corner → one weight becomes zero.

L2 constrains to a **circle**. A circle has **no corners**, so weights rarely become zero.

# 🟥1.4 Explain how you can use the same models (Linear regression, lasso, etc.) but make it possible to fit nonlinear dependencies.

нужно создать новые признаки, которые являются нелинейными функциями от исходных, а затем применить обычную линейную регрессию к этим новым признакам.

## How to fit nonlinear dependencies with linear models

### 1. The problem

Linear regression produces predictions of the form:

$$
\hat{y} = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \dots
$$

This is a straight line (2D), plane (3D), or hyperplane (higher dimensions).  
If the true relationship is curved (e.g., $y = x^2$), a linear model will fail.

### 2. The solution: create new features

Keep the model **linear in parameters**, but add columns to $X$ that are nonlinear functions of the original features.

**Example  $y = x^2$:**

Create the feature matrix:

$$
X = \begin{pmatrix} 1 & x & x^2 \end{pmatrix}
$$

The model becomes:

$$
\hat{y} = \beta_0 + \beta_1 x + \beta_2 x^2
$$

This is **polynomial regression**. The model is linear in $\beta$, so we solve it the same way (OLS, Ridge, Lasso).

### 3. Other examples of nonlinear features

| Original feature $x$ | New features | Can model |
| :--- | :--- | :--- |
| $x$ | $x, x^2, x^3$ | Polynomial relationships |
| $x$ | $\sin(x), \cos(x)$ | Periodic relationships |
| $x$ | $\log(x), \sqrt{x}, e^x$ | Logarithmic, power, exponential |
| $x_1, x_2$ | $x_1 \cdot x_2$ | Feature interactions |

# 🟩4.3 Define the R squared (R2) coefficient and implement a function to calculate it.

```python
class Linear:
    def __init__(self, learning_rate, n_epochs, random_state=21, penalty='none', 
                 alpha=1, l1_ratio=0.5, grad_clip=1.0):
        self.learning_rate = learning_rate
        self.n_epochs = n_epochs
        self.beta = None
        self.random_state = random_state
        self.penalty = penalty
        self.alpha = alpha
        self.l1_ratio = l1_ratio
        self.grad_clip = grad_clip

    def _regularization_gradient(self):
        grad_reg = np.zeros(len(self.beta))

        if self.penalty == 'Lasso':
            grad_reg[1:] = self.alpha * np.sign(self.beta[1:])
        elif self.penalty == 'Ridge':
            grad_reg[1:] = 2 * self.alpha * self.beta[1:]
        elif self.penalty == 'ElasticNet':
            grad_reg[1:] = (self.alpha * self.l1_ratio * np.sign(self.beta[1:]) + 
                           2 * self.alpha * (1 - self.l1_ratio) * self.beta[1:])
        return grad_reg
    
    def fit(self, X, y):
        if hasattr(X, 'values'):
            X = X.values
        if hasattr(y, 'values'):
            y = y.values

        if X is None or y is None:
            raise ValueError('X и y не могут быть None')
        if len(X) == 0:
            raise ValueError('Длина X не должна равняться 0')
        if len(X) != len(y):
            raise ValueError('Длины X и y не равны')
        
        ones = np.ones((X.shape[0], 1))
        X = np.hstack([ones, X])
        self.beta = np.zeros(X.shape[1])

        n = X.shape[0]
        np.random.seed(self.random_state) 
    
        for epoch in range(self.n_epochs):
            indexes = np.random.permutation(n)
            X = X[indexes]
            y = y[indexes]

            for i in range(n):
                X_batch = X[i:i+1]
                y_batch = y[i]

                y_pred = X_batch @ self.beta
                error = y_pred - y_batch

                grad_reg = self._regularization_gradient()
                gradient = 2 * X_batch.T @ error + grad_reg.flatten()
                
                # Клиппинг градиента (ограничиваем максимальное значение)
                grad_norm = np.linalg.norm(gradient)
                if grad_norm > self.grad_clip:
                    gradient = gradient * (self.grad_clip / grad_norm)
                
                self.beta = self.beta - self.learning_rate * gradient
                
                # Проверка на NaN/Inf
                if np.any(np.isnan(self.beta)) or np.any(np.isinf(self.beta)):
                    print(f"Warning: NaN/Inf at epoch {epoch}, iter {i}")
                    return self

        return self
    
    def predict(self, X):
        if hasattr(X, 'values'):
            X = X.values
        ones = np.ones((X.shape[0], 1))
        X = np.hstack([ones, X])
        return X @ self.beta

    def r2_score(self, y, y_pred):
        y_mean = np.mean(y)
        ss_res = np.sum((y - y_pred) ** 2)
        ss_tot = np.sum((y - y_mean) ** 2)
        if ss_tot == 0:
            return 1.0
        return 1 - (ss_res / ss_tot)

    def mae(self, y_true, y_pred):
        return np.mean(np.abs(y_true - y_pred))
    
    def rmse(self, y_true, y_pred):
        return np.sqrt(np.mean((y_true - y_pred) ** 2))
```

```python
# Обучаем модель
model = Linear(learning_rate=0.01, n_epochs=100, random_state=21)
model.fit(X_train, y_train)

y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

mae_train = model.mae(y_train, y_train_pred)
rmse_train = model.rmse(y_train, y_train_pred)
r2_train = model.r2_score(y_train, y_train_pred)

mae_test = model.mae(y_test, y_test_pred)
rmse_test = model.rmse(y_test, y_test_pred)
r2_test = model.r2_score(y_test, y_test_pred)


my_metrics = {
    'Train MAE':mae_train,
    'Test MAE': mae_test,
    'Train RMSE': rmse_train,
    'Test RMSE': rmse_test,
    'Train R2': r2_train,
    'Test R2': r2_test
}
```

# 🟪6.1 First, write several examples of why and where feature normalization is mandatory and vice versa.

Нормализация необходима, если у нас много плюс минус одинаковых по важности числовых признаков, и чтобы модель показывала хорошие результаты нам надо привести это к одному формату. 
Например в диапозоне от 0 до 1. 
Если у нас есть кол-во комнат и цена за квадратный метр. Это два достаточно важных признака, без нормализации, вес будет распределяться неправильно. Модель будет считать, что цена важнее, так как числа больше. Это неправильно.

# 🟪6.2 Let's consider the first of the classical normalization methods — MinMaxScaler. Write a mathematical formula for this method.



$$
x_{norm} = \frac{x - x_{min}}{x_{max} - x_{min}}
$$
​
 
​


```python
class MinMax:
    def __init__(self):
        self.min_ = None
        self.max_ = None
        pass
    def fit(self, X):
        if hasattr(X, 'values'):
            X = X.values
        if len(X) == 0:
            raise ValueError("Пустой массив")

        self.min_ = np.min(X, axis=0)
        self.max_ = np.max(X, axis=0)
        return self
    def transform(self, X):
        if hasattr(X, 'values'):
            X = X.values
        if X is None:
            raise ValueError("Матрица X пустая")
        if len(X) == 0:
            raise ValueError("Кол-во строк 0")
        if self.min_ is None or self.max_ is None:
            raise ValueError("Сначала вызовите fit()")
        if X.shape[1] != len(self.min_) or X.shape[1] != len(self.max_):
            raise ValueError("оличество столбцов X не совпадает с количеством признаков в fit")
        numerator = X - self.min_
        denominator = self.max_ - self.min_
        denominator[denominator == 0] = 1
        zero_denom = denominator == 0
        if np.any(zero_denom):
            denominator[zero_denom] = 1
            numerator[:, zero_denom] = 0
        X_normalised = numerator/ denominator
        return X_normalised
    def fit_transform(self, X):
        self.fit(X)
        return self.transform(X)
        
```

$$
x_{std} = \frac{x - \mu}{\sigma}
$$

$$
\mu = \frac{1}{n} \sum_{i=1}^{n} x_i
$$

$$
\sigma = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (x_i - \mu)^2}
$$


x — исходное значение

μ (мю) — среднее арифметическое всех значений этого признака

σ (сигма) — стандартное отклонение всех значений этого признака

x std — стандартизированное значение

```python
class StandardScalerMy:
    def __init__(self):
        self.mean_ = None
        self.std_ = None
    
    def fit(self, X):
        if hasattr(X, 'values'):
            X = X.values
        if len(X) == 0:
            raise ValueError("Пустой массив")
        
        self.mean_ = np.mean(X, axis=0)
        self.std_ = np.std(X, axis=0)
        return self
    
    def transform(self, X):
        if hasattr(X, 'values'):
            X = X.values
    
        if X is None:
            raise ValueError("Матрица X пустая")
        if len(X) == 0:
            raise ValueError("Кол-во строк 0")
        if self.mean_ is None or self.std_ is None:
            raise ValueError("Сначала вызовите fit()")
        if X.shape[1] != len(self.mean_):
            raise ValueError("Количество столбцов X не совпадает с количеством признаков в fit")
        
        numerator = X - self.mean_
        denominator = self.std_
        
        zero_std = denominator == 0
        if np.any(zero_std):
            denominator[zero_std] = 1
            numerator[:, zero_std] = 0
        X_standardised = numerator / denominator
        return X_standardised
    
    def fit_transform(self, X):
        self.fit(X)
        return self.transform(X)
```

# ⭐️11.3 It will also be a useful exercise to implement a linear regression algorithm with batch and mini-batch training or analytical solution (as mentioned in 4.1).

```python
class MyLinearRegression():
    def __init__(self, method='analytical', learning_rate=0.01, n_epochs=1000, batch_size=None, random_state=21):
        self.method = method
        self.learning_rate = learning_rate
        self.n_epochs = n_epochs
        self.batch_size = batch_size
        self.random_state = random_state
        self.weights_ = None
        self.bias_ = None
        
    def _add_bias(self, X):
        return np.c_[np.ones(X.shape[0]), X]
    
    def fit(self, X, y):
        X = np.array(X)
        y = np.array(y)
        
        if self.method == 'analytical':
            self._fit_analytical(X, y)
        elif self.method in ['batch', 'mini-batch']:
            self._fit_gradient_descent(X, y)
        else:
            raise ValueError("Method must be 'analytical', 'batch', or 'mini-batch'")
            
        return self
    
    def _fit_analytical(self, X, y):
        X_b = self._add_bias(X)
        self.weights_ = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y
        self.bias_ = self.weights_[0]
        self.weights_ = self.weights_[1:]
        
    def _fit_gradient_descent(self, X, y):
        np.random.seed(self.random_state)
        n_samples, n_features = X.shape
        
        self.weights_ = np.random.normal(0, 0.01, n_features)
        self.bias_ = 0
        
        if self.method == 'batch' or self.batch_size is None:
            batch_size = n_samples
        else:  
            batch_size = min(self.batch_size, n_samples)
            
        for epoch in range(self.n_epochs):
            if self.method == 'mini-batch':
                indices = np.random.permutation(n_samples)
                X_shuffled = X[indices]
                y_shuffled = y[indices]
            else:
                X_shuffled = X
                y_shuffled = y
                
            for i in range(0, n_samples, batch_size):
                X_batch = X_shuffled[i:i+batch_size]
                y_batch = y_shuffled[i:i+batch_size]
                
                y_pred = X_batch @ self.weights_ + self.bias_
                
                dw = (2 / len(X_batch)) * X_batch.T @ (y_pred - y_batch)
                db = (2 / len(X_batch)) * np.sum(y_pred - y_batch)
                
                self.weights_ -= self.learning_rate * dw
                self.bias_ -= self.learning_rate * db
                
    def predict(self, X):
        X = np.array(X)
        return X @ self.weights_ + self.bias_
```

```python
model_analytical = MyLinearRegression(method='analytical')
model_analytical.fit(X_train, y_train)
y_pred_train_analytical = model_analytical.predict(X_train)
y_pred_test_analytical = model_analytical.predict(X_test)

metrics_table.append({
    'Model': 'Linear (Analytical)',
    'Train MAE': mean_absolute_error(y_train, y_pred_train_analytical),
    'Test MAE': mean_absolute_error(y_test, y_pred_test_analytical),
    'Train RMSE': root_mean_squared_error(y_train, y_pred_train_analytical),
    'Test RMSE': root_mean_squared_error(y_test, y_pred_test_analytical),
    'Train R2': r2_score(y_train, y_pred_train_analytical),
    'Test R2': r2_score(y_test, y_pred_test_analytical)
})

model_batch = MyLinearRegression(
    method='batch', 
    learning_rate=0.01, 
    n_epochs=1000,
    random_state=21
)
model_batch.fit(X_train, y_train)
y_pred_train_batch = model_batch.predict(X_train)
y_pred_test_batch = model_batch.predict(X_test)

metrics_table.append({
    'Model': 'Linear (Batch GD)',
    'Train MAE': mean_absolute_error(y_train, y_pred_train_batch),
    'Test MAE': mean_absolute_error(y_test, y_pred_test_batch),
    'Train RMSE': root_mean_squared_error(y_train, y_pred_train_batch),
    'Test RMSE': root_mean_squared_error(y_test, y_pred_test_batch),
    'Train R2': r2_score(y_train, y_pred_train_batch),
    'Test R2': r2_score(y_test, y_pred_test_batch)
})

model_mini = MyLinearRegression(
    method='mini-batch', 
    learning_rate=0.01, 
    n_epochs=1000,
    batch_size=32,
    random_state=21
)
model_mini.fit(X_train, y_train)
y_pred_train_mini = model_mini.predict(X_train)
y_pred_test_mini = model_mini.predict(X_test)

metrics_table.append({
    'Model': 'Linear (Mini-batch GD)',
    'Train MAE': mean_absolute_error(y_train, y_pred_train_mini),
    'Test MAE': mean_absolute_error(y_test, y_pred_test_mini),
    'Train RMSE': root_mean_squared_error(y_train, y_pred_train_mini),
    'Test RMSE': root_mean_squared_error(y_test, y_pred_test_mini),
    'Train R2': r2_score(y_train, y_pred_train_mini),
    'Test R2': r2_score(y_test, y_pred_test_mini)
})

```
